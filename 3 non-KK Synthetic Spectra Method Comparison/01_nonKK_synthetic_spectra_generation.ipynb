- Main parameters: CODE_DIR.These variables control input/output paths, experimental conditions, frequency ranges, or plotting behavior.

```python
from pathlib import Path
import sys
NOTEBOOK_DIR = Path.cwd()
CODE_DIR = NOTEBOOK_DIR if (NOTEBOOK_DIR / "nonkk_common.py").exists() else NOTEBOOK_DIR / "3 non-KK Synthetic Spectra Method Comparison"
if not (CODE_DIR / "nonkk_common.py").exists():
    raise FileNotFoundError("nonkk_common.py was not found. Run this notebook from the current folder or from the Github code/ENG version root directory.")
if str(CODE_DIR) not in sys.path:
    sys.path.insert(0, str(CODE_DIR))
from nonkk_common import *
print("CODE_DIR:", CODE_DIR)
print("DATA_ROOT:", DATA_ROOT)
```

### Cell 02 - Main Workflow Execution

This cell runs the main workflow or intermediate data-organization steps of the current notebook.

- Purpose: Generation of double Cole and Rct-ZARC non-KK synthetic spectra.

```python
# Parameter and hyperparameter settings
# Frequency range and number of points
# Noise-condition settings
# Sparse-sampling condition settings
# Frequency-limitation or endpoint-truncation settings
generated = {}
for model_key in MODEL_SPECS:
    generated[model_key] = generate_one_model_dataset(model_key)
    print(model_key, "conditions:", list(generated[model_key]["inputs"].keys()))
```

### Cell 03 - Main Workflow Execution

This cell runs the main workflow or intermediate data-organization steps of the current notebook.

- Purpose: Generation of double Cole and Rct-ZARC non-KK synthetic spectra.

```python
# Save results
ensure_dir(DATA_DIR)
saved_workbooks = {}
for model_key, dataset in generated.items():
    saved_workbooks[model_key] = write_dataset_workbook(dataset)
    print("saved workbook:", saved_workbooks[model_key])
```

### Cell 04 - Plotting and Figure Export

This cell plots Nyquist, Bode, or method-comparison figures and saves them with a consistent naming rule.

- Purpose: Generation of double Cole and Rct-ZARC non-KK synthetic spectra.

```python
# Figure results are saved to the specified output directory
# Model settings
ensure_dir(DATA_FIG_DIR)
saved_figures = []
for model_key, dataset in generated.items():
    saved_figures.extend(plot_dataset(dataset))
print("number of saved figures:", len(saved_figures))
print("figure folder:", DATA_FIG_DIR)
```

### Cell 05 - Main Workflow Execution

This cell runs the main workflow or intermediate data-organization steps of the current notebook.

- Purpose: Generation of double Cole and Rct-ZARC non-KK synthetic spectra.

```python
# Table results are saved to the specified output directory
summary_rows = []
for model_key, dataset in generated.items():
    for cond, df in dataset["inputs"].items():
        summary_rows.append({
            "model": model_key,
            "condition": cond,
            "n_points": len(df),
            "f_min_Hz": df["freq"].min(),
            "f_max_Hz": df["freq"].max(),
            "workbook": str(saved_workbooks[model_key]),
        })
summary = pd.DataFrame(summary_rows)
summary_path = DATA_DIR / "nonKK_dataset_summary.csv"
summary.to_csv(summary_path, index=False, encoding="utf-8-sig")
display(summary)
print("saved:", summary_path)
```
